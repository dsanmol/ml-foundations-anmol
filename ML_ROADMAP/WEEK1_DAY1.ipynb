```python
data=pd.read_csv("customer_segmentation.csv")
```

```python
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data['Dt_Customer']=pd.to_datetime(data['Dt_Customer'],format='%d-%m-%Y')
```

```python
data
```

```python
data.shape
```

```python
data['Age']=2026-data['Year_Birth']
```

```python
data=data[(data['Age']<90) & (data['Income']<100000)]
```

```python
data.shape
```

```python
data['Total_Mnt'] = data['MntWines'] + data['MntFruits'] + data['MntMeatProducts'] + data['MntFishProducts'] + data['MntSweetProducts'] + data['MntGoldProds']
data['Total_Purchases'] = data['NumWebPurchases'] + data['NumCatalogPurchases'] + data['NumStorePurchases']
data['Children'] = data['Kidhome'] + data['Teenhome']
data['Family_Size'] = data['Marital_Status'].replace({'Married': 2, 'Together': 2, 'Single': 1, 'Divorced': 1, 'Widow': 1, 'Alone': 1, 'Absurd': 1, 'YOLO': 1}) + data['Children']
```

```python
cols = ['Age', 'Income', 'Total_Mnt', 'Total_Purchases','Family_Size','Recency','Children']
print(data[cols].describe().T)
```

```python
cols = ['Age', 'Income', 'Total_Mnt', 'Total_Purchases','Family_Size','Recency','Children']
print(data[cols].describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']])
```

```python
#Simple pythonic way to normalize features
def norm(arr):
    mean = np.mean(arr)
    std = np.std(arr)

    normalized = []

    for nm in arr:
        z = (nm - mean) / std
        normalized.append(z)

    return np.array(normalized)   
```

```python
norm(arr)
```

```python
# Numpy vectorized way to normalize features
def norm(arr):
    mean = np.mean(arr)
    std = np.std(arr)
    return (arr - mean) / std
```

```python
def norm_col(arr):
    mean = np.mean(arr, axis=0)
    std = np.std(arr, axis=0)
    normalized = (arr - mean) / std
    return normalized
```
